```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../data/l2_conventional_data_202611042047.csv")
print(df.info())
```

```python
cs_ola = df[["credit_score", "original_loan_amt"]]
print(cs_ola.iloc[:10])
```

```python
plt.hist(
    df["credit_score"].dropna(),
    bins=30,
    color="skyblue",  # Bar fill color
    edgecolor="black",  # Border color
    linewidth=1.2,  # Border thickness
)
plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.show()
```

```python
plt.hist(
    df["original_loan_amt"].dropna(),
    bins=30,
    color="skyblue",  # Bar fill color
    edgecolor="black",  # Border color
    linewidth=1.2,  # Border thickness
)
plt.title("Original Loan Amount Distribution")
plt.xlabel("Original Loan Amount ($)")
plt.show()
```

```python
plt.scatter(x = df["credit_score"],
            y = df["original_loan_amt"],
            color = "skyblue",
            alpha = 0.2)
plt.title("Credit Score vs. Original Loan Amount")
plt.xlabel("Credit Score")
plt.ylabel("Original Loan Amount ($)")
plt.show()
```

```python
plt.hist(x = df["unemployment_rate_origination"],
         bins = 15,
         color = "skyblue",
         edgecolor = "black",
         linewidth = 1.2)
plt.xlabel("Unemployment Rate for Loan Geography at Origination (%)")
plt.show()
```

```python
avg_loan = (
    df.groupby("unemployment_rate_origination")["original_loan_amt"]
    .mean()
    .reset_index()
)

plt.bar(
    x = avg_loan["unemployment_rate_origination"],
    height=avg_loan["original_loan_amt"],
    edgecolor = "black",
    color = "skyblue",
    width = 0.08,
    linewidth = 1.2
)
plt.xlabel("Unemployment Rate for Loan Geography at Origination (%)")
plt.ylabel("Average Original Loan Amount ($)")
plt.show()
```
